```python
import os
import json



def print_video_ids_from_clean_files(file_dir):

    file_ids = []
    for filename in os.listdir(file_dir):
        if filename.endswith(".json"):
            with open(os.path.join(file_dir, filename), "r", encoding="utf-8") as f:
                data = json.load(f)
            append_id = data.get("video_id")
            if append_id:
                file_ids.append(append_id)

    return file_ids


all_ids = print_video_ids_from_clean_files("output/clean_files")

for video_id in range(5):
    print(all_ids[video_id])
```

```python
import yt_dlp
import json

def extract_video_info(video_ids):
    ydl_opts = {
        "quiet": True
    }

    need_results_keys = ["id", "url", "title", "channel", "upload_date", "tags", "view_count", "description"]

    valid_results = []

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        for vid in video_ids:
            url = f"https://www.youtube.com/watch?v={vid}"
            try:
                info = ydl.extract_info(url, download=False)
                info = {k: info.get(k) for k in need_results_keys}
                valid_results.append(info)



            except Exception as e:
                error_msg = str(e)

                if "Private video" in error_msg:
                    print(f"Skipping private video: {vid}")
                elif "Video unavailable" in error_msg:
                    print(f"Skipping unavailable video: {vid}")
                else:
                    print(f"Skipping {vid} due to error: {error_msg}")

                continue


    return valid_results





video_id = "4zFyqwc_kHQ"
video_info = extract_video_info([video_id])
print(json.dumps(video_info, indent=2))
```
