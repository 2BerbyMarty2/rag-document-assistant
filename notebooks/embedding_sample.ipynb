```python
data_path = "final_data/clean_transcripts"
```

```python
import json
import os
```

```python
files = []
```

```python
for i in os.listdir(data_path):
    a = i.split('.')[0]
    files.append(a)
```

```python
file_path = os.path.join(data_path, files[3] + ".json")

with open(file_path, "r") as f:
    file_data = json.load(f)
```
